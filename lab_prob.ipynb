```python
import numpy as np
import pandas as pd
```

```python
def pmf_through_sample(n,p):

    rng = np.random.default_rng(seed=42)
    samples = rng.binomial(n=1, p=p, size=n)

    zeros=np.sum(samples==0)
    ones=np.sum(samples==1)
    prob_noclick=zeros/n
    prob_click=ones/n
    possible_outcomes=np.array([0,1])
    Theory_pmf = np.array([1-p,p])
    pmf=np.array([prob_noclick,prob_click])
    print("pmf",pmf)
    print("Theory_pmf",Theory_pmf)
    return abs(pmf[1]-Theory_pmf[1]),pmf@possible_outcomes,Theory_pmf@possible_outcomes
sizes=[10,100,1000,10000]

import pandas as pd

data = []

for i in sizes:
    diff, expectation,Theory = pmf_through_sample(i, 0.3)
    data.append({
        "Sample Size": i,
        "Difference": diff,
        "exp_Expectation": expectation,
        "Theory_expectation":Theory
    })

df = pd.DataFrame(data)

print(df)

def plot(pmf,Theory_pmf):

    import matplotlib.pyplot as plt
    x = np.arange(len(pmf))
    width = 0.35  

    fig, ax = plt.subplots(figsize=(8, 5))

    bars1 = ax.bar(x - width/2, Theory_pmf, width, label='Theoretical PMF', color='#1f77b4', alpha=0.85)
    bars2 = ax.bar(x + width/2, pmf, width, label='Empirical PMF (N=20)', color='#ff7f0e', alpha=0.85)


    ax.set_ylabel('Probability', fontsize=11)
    ax.set_title('Bernoulli(0.30): Theoretical vs. Empirical PMF', fontsize=13, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(['X = 0\n(No Click)', 'X = 1\n(Click)'], fontsize=10)
    ax.set_ylim(0, 1.0)
    ax.legend(fontsize=10)
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.bar_label(bars1, fmt='%.2f', padding=3)
    ax.bar_label(bars2, fmt='%.2f', padding=3)

    plt.tight_layout()
    plt.show()


    

    
    
```
