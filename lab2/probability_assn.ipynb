```python
import random
print("assaingment")
def ntoss(n):
    headc=0
    tailc=0
    for _ in range(n):
        toss=random.choice(["head","tail"])
        if toss=="head":
            headc+=1
        else:
            tailc+=1
    return headc,tailc
def tossres(n):
    res=ntoss(n)
    print("heads :",res[0])
    print("tails :",res[1])
    print("prob of head :",res[0]/sum(res))
    print("prob of tail :",res[1]/sum(res))
tossres(100)

```

```python
print("for 10 tosses \n")
print(tossres(10))
print("for 100 tosses \n")
print(tossres(100))
print("for 1000 tosses \n")
print(tossres(1000))
print("for 10000 tosses \n")
print(tossres(10000))
```

```python
def dice(n):
    res=[0,0,0,0,0,0]
    for _ in range(n):
        roll=random.choice([1,2,3,4,5,6])
        res[roll-1]+=1
    return res
print(dice(6000))
```

```python
roll_ten_t=dice(10000)
x=roll_ten_t[0]+roll_ten_t[5]
y=10000 - sum(roll_ten_t[1:5]) 
if x==y:
    print("disjoint")
else:
    print("joint")
```

```python
ab=dice(1000)
prob_A = (ab[1] + ab[3] + ab[5]) / sum(ab)
prob_B = sum(ab[3:]) / sum(ab)

prob_A_u_B = (sum(ab[3:]) + ab[1]) / sum(ab)
prob_A_in_B = (ab[3] + ab[5]) / sum(ab)

if prob_A_u_B == prob_A + prob_B - prob_A_in_B:
    print("True")
else:
    print("False")
```

```python
def mul_rule(n):
    head=0
    even=0
    head_even=0
    even_set={2,4,6}
    for _ in range(n):
        toss=random.choice(["head","tail"])
        die=random.choice([1,2,3,4,5,6])
        if toss=="head":
            head+=1
        if die in even_set:
            even+=1
        if toss=="head" and die in even_set:
            head_even+=1
    head=head/n
    even=even/n
    head_even=head_even/n
    if head_even==(head * even):
        print("mul rule is satisfies")
    else:
        print("not satisfied")
mul_rule(10000)
             
    

```
